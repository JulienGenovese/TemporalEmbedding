# Output embedding - `simple_delta`

Notebook rapido per ispezionare gli embedding prodotti da
`pred_simple_delta_embeddings.csv` e verificare se la separazione tra i
cluster emerge anche nello spazio latente del modello.


```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from sklearn.decomposition import PCA

plt.rcParams['figure.figsize'] = (9, 5)
plt.rcParams['axes.grid'] = True
pd.set_option('display.max_columns', 60)
```

```python
def find_repo_root(start: Path) -> Path:
    for p in [start, *start.parents]:
        if (p / 'data').is_dir() and (p / 'config.toml').exists():
            return p
    return start


ROOT = find_repo_root(Path.cwd())
EMB = ROOT / 'data' / 'output' / 'pred_simple_delta_embeddings.csv'
PRED = ROOT / 'data' / 'transactions_simple_delta_pred.csv'

hemb = pd.read_csv(EMB)
df_pred = pd.read_csv(PRED)

print('Embeddings:', EMB)
print('Prediction input:', PRED)
print('Shape embeddings:', hemb.shape)
print('Shape pred input:', df_pred.shape)

display(hemb.head())
df_pred.head()
```

```python
colors_map = {
    'veloce': 'tab:green',
    'medio': 'tab:blue',
    'lento': 'tab:orange',
    'lentissimo': 'tab:red',
}

emb_cols = [c for c in hemb.columns if c.startswith('emb_')]
if not emb_cols:
    raise ValueError('Nessuna colonna emb_* trovata nel file embeddings.')

hemb_last = (
    hemb
    .sort_values(['client_id', 'window_index'])
    .groupby('client_id', as_index=False)
    .last()
)

map_client_id = df_pred[['client_id', 'cluster']].drop_duplicates()
plot_df = hemb_last.merge(map_client_id, on='client_id', how='left').dropna(subset=['cluster']).copy()

plot_df[['client_id', 'cluster', 'window_index']].head()
```

```python
summary = plot_df.groupby('cluster').agg(
    n_clients=('client_id', 'nunique'),
    avg_window_index=('window_index', 'mean'),
)
summary
```

```python
pca = PCA(n_components=2, random_state=42)
emb_2d = pca.fit_transform(plot_df[emb_cols])
evr = pca.explained_variance_ratio_

fig, ax = plt.subplots(figsize=(8, 6))
for cluster, group in plot_df.groupby('cluster'):
    idx = group.index.to_numpy()
    ax.scatter(
        emb_2d[idx, 0],
        emb_2d[idx, 1],
        alpha=0.55,
        s=18,
        color=colors_map.get(cluster, 'gray'),
        label=cluster,
    )
ax.set(
    xlabel=f'PCA 1 ({evr[0]:.0%})',
    ylabel=f'PCA 2 ({evr[1]:.0%})',
    title='Proiezione PCA degli embedding dei clienti',
)
ax.legend(title='Cluster')
plt.tight_layout()
```

```python
centroids = (
    pd.DataFrame(emb_2d, index=plot_df.index, columns=['pc1', 'pc2'])
    .assign(cluster=plot_df['cluster'].values)
    .groupby('cluster')[['pc1', 'pc2']]
    .mean()
    .sort_index()
)
centroids.round(3)
```
